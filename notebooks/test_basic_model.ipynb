```python
import sys
from pathlib import Path

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "src" / "preprocessing" / "align.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the project checkout.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data import *
from src.model import *
```

```python
train, val, test = get_dataset()
```

```python
train_az = get_language_data(train, "en", "az")[:1000]
train_be = get_language_data(train, "en", "be")[:1000]

val_az = get_language_data(val, "en", "az")[:100]
val_be = get_language_data(val, "en", "be")[:100]

test_az = get_language_data(test, "en", "az")[:100]
test_be = get_language_data(test, "en", "be")[:100]
```

```python
en_vocab = Vocabulary([s[0] for s in train_az + val_az + test_az] + [s[0] for s in train_be + val_be + test_be])
az_vocab = Vocabulary([s[1] for s in train_az + val_az + test_az])
be_vocab = Vocabulary([s[1] for s in train_be + val_be + test_be])
```

```python
model_az = Seq2SeqModel(
    src_vocab=en_vocab,
    tgt_vocab=az_vocab
)
```

```python
model_az.fit(pairs=train_az,epochs=10)
```
